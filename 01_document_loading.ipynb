```python
from pathlib import Path
from langchain_community.document_loaders import PyPDFLoader

# Path containing all the PDF documents used as our knowledge base
documents_path = Path("data/documents")

# Find every PDF file inside the documents folder
pdf_files = list(documents_path.glob("*.pdf"))

print("Number of PDF files:", len(pdf_files))

# Store all pages from all PDFs in one list
all_documents = []

# Load each PDF and add its pages to the document list
for pdf_file in pdf_files:
    loader = PyPDFLoader(str(pdf_file))
    docs = loader.load()

    # Add metadata about the source PDF
    for doc in docs:
        doc.metadata["source_file"] = pdf_file.name

    all_documents.extend(docs)

print("Total pages loaded:", len(all_documents))
```

```python
from collections import Counter

# Count how many pages belong to each source PDF
page_counts = Counter(
    doc.metadata["source_file"]
    for doc in all_documents
)

for filename, count in page_counts.items():
    print(f"{filename}: {count} pages")
```
