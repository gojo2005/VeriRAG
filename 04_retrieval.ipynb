# Retrieval

```python
# We need the same embedding model that was used
# when we originally stored our documents in ChromaDB.
# IMPORTANT:
# The query must be embedded using the same model
# used for the document embeddings.

from langchain_huggingface import HuggingFaceEmbeddings

embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

print("Embedding model loaded successfully.")
```

```python
# We are NOT creating a new database here.
# We are connecting to the ChromaDB that we created
# in 03_embedding.ipynb.

from langchain_chroma import Chroma

vectorstore = Chroma(
    persist_directory="chroma_db",
    embedding_function=embedding_model
)

print("Connected to existing ChromaDB successfully.")
```

```python
# This is our first real retrieval step.
# Chroma will:
# 1. Convert our question into an embedding.
# 2. Compare it with the stored document embeddings.
# 3. Return the most semantically similar chunks.

query = "How does the Transformer architecture use attention?"

results = vectorstore.similarity_search(
    query,
    k=3  # k tells how many results we want for the query asked by user
)

print("Query:", query)
print("Number of retrieved chunks:", len(results))

for i, result in enumerate(results):

    print(f"\n{'=' * 70}")
    print(f"RESULT {i + 1}")
    print(f"{'=' * 70}")

    print(result.page_content[:700])

    print("\nSource:", result.metadata.get("source_file"))
    print("Page:", result.metadata.get("page"))
```
